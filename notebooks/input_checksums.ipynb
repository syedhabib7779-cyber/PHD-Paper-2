# Input checksums (SHA-256)
Attach the same datasets as the main notebook (CICIDS2017, cicdos2017, and the 2018 files if attached). Internet ON. No GPU. Save & Run All, then download input_checksums.csv.

```
import hashlib, os, glob, urllib.request, pandas as pd
def sha(p):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        for b in iter(lambda: f.read(1 << 20), b""): h.update(b)
    return h.hexdigest()
rows = []
for p in sorted(glob.glob("/kaggle/input/**/*", recursive=True)):
    if os.path.isfile(p) and p.lower().endswith((".csv", ".parquet")):
        rows.append({"file": p.replace("/kaggle/input/", ""), "bytes": os.path.getsize(p), "sha256": sha(p)}); print(rows[-1])
BASE = "https://cse-cic-ids2018.s3.ca-central-1.amazonaws.com/Processed%20Traffic%20Data%20for%20ML%20Algorithms/"
for f in ("Thursday-15-02-2018_TrafficForML_CICFlowMeter.csv", "Friday-16-02-2018_TrafficForML_CICFlowMeter.csv"):
    try:
        d = "/tmp/" + f; urllib.request.urlretrieve(BASE + f, d)
        rows.append({"file": "s3://cse-cic-ids2018/Processed Traffic Data for ML Algorithms/" + f, "bytes": os.path.getsize(d), "sha256": sha(d)}); print(rows[-1])
    except Exception as e:
        print("download failed", f, e)
pd.DataFrame(rows).to_csv("/kaggle/working/input_checksums.csv", index=False)
print(pd.DataFrame(rows).to_string(index=False))
```
